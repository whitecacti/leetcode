```python
def check_for_target(nums, target):
    left = 0
    right = len(nums) - 1

    while left < right:
        # curr is the current sum
        curr = nums[left] + nums[right]
        if curr == target:
            print(left, right)
            print(nums[left], nums[right])
            return True
        if curr > target:
            right -= 1
        else:
            left += 1
    
    return False
```

```python
check_for_target([1,2,3,4,5,6],8)
```

```python
def isSubsequence(s: str, t: str) -> bool:
    i = j = 0
    while i < len(s) and j < len(t):
        if s[i] == t[j]:
            i += 1
        j += 1

    return i == len(s)
```

```python
isSubsequence('ace','abcde')
```

```python
def func(nums):
    left = 0
    right = len(nums)-1
    while left < len(nums):
        if nums[left] == nums[right]:
            return [nums[left], nums[left] + 1]
        
        left += 1
```

```python
from typing import List

def reverseString(s: List[str]) -> None:

    """
    Do not return anything, modify s in-place instead.
    """
    left = 0
    right = len(s) - 1
    while left <= right:
        leftL = s[left]
        rightL = s[right]
        s[left] = rightL
        s[right] = leftL
        left+=1
        right-=1

    return s
```

```python
reverseString(['o','n','e','s'])
```

```python
from typing import List
def sortedSquares(nums: List[int]) -> List[int]:
    start = 0
    end = len(nums) - 1
    array = list(range(end +1 ))
    while start <= end:
        if nums[start] ** 2 < nums[end] ** 2:
            array[end-start] = nums[end] ** 2
            end-=1
        else:
            array[end-start] = nums[start] ** 2
            start+=1

    return array

# Need 2 points, going from out in as thats the order thats sorted.
# Once we are in the middle we are done
# As we loop through we are adding whichever side is greater to array first
```
