```python
# LinkedList class

class LinkedList():
    # def __init__(self, head: Node, size):
    def __init__(self):
        self.head = Node() 
        self.size = 0

    # prepend(val)
    def prepend(self, val):
        # return Node(val).next 
        placeholder = self.head.next
        self.head.next = Node(val, placeholder)
        self.size+=1

    # get(index)
    def get(self, index):
        if index < self.size:
            placeholder = self.head.next
            for i in range(index):
                placeholder = placeholder.next

            return placeholder.value
        else:
            return None
        
    # add(val)
    def add(self, val):
        placeholder = self.head
        while placeholder.next is not None:
            placeholder = placeholder.next
        placeholder.next = Node(val)
        self.size+=1

    # size()
    def size(self):
        return self.size

    #remove(index)
    def remove(self, index):
        placeholder = self.head.next
        for i in range(index - 1):
            placeholder = placeholder.next

        placeholder.next = placeholder.next.next
        self.size-=1

    #insert(index, val)
    def insert(self, index, val):
        placeholder = self.head.next
        for i in range(index - 1):
            placeholder = placeholder.next

        temp = placeholder.next
        placeholder.next = Node(val, temp)
        self.size+=1
```

```python
ll = LinkedList()
ll.add(5)
ll.add(10)
ll.remove(1)
ll.add(15)
ll.prepend(100)
ll.insert(1,50)
```

```python
# Node class

class Node():
    def __init__(self, val = None, next = None, prev = None):
        self.value = val
        self.next = next
        self.prev = prev

# Double LinkedList class

class DoubleLinkedList():
    # def __init__(self, head: Node, size):
    def __init__(self):
        self.head = Node() 
        self.tail = Node()
        self.head.next = self.tail
        self.tail.prev = self.head
        self.size = 0

    # prepend(val)
    def prepend(self, val):
        # return Node(val).next 
        placement = Node(val)
        placeholder_head = self.head.next

        self.head.next = placement
        placeholder_head.prev = placement

        placement.prev = self.head
        placement.next = placeholder_head
        
        self.size+=1

    # get(index)
    def get(self, index):
        if index < self.size:
            placeholder = self.head.next
            for i in range(index):
                placeholder = placeholder.next

            return placeholder.value
        else:
            return None
        
    # # add(val)
    # head(n) -> (p)(5)(n) -> (p)tail
    # head(n) -> (p)(5)(n) -> (p)(10)(n) -> (p)tail
    # head(n) -> (p)(5)(n) -> (p)(10)(n) -> (p)(15)(n) -> (p)tail
    def add(self, val):
        placement = Node(val)
        placeholder = self.tail.prev

        self.tail.prev = placement
        placeholder.next = placement

        placement.next = self.tail
        placement.prev = placeholder

        self.size+=1

    # size()
    def size(self):
        return self.size

    #remove(index)
    def remove(self, index):
        placeholder = self.head.next
        for i in range(index - 1):
            placeholder = placeholder.next

        temp = placeholder.next.next.prev
        placeholder.next = placeholder.next.next
        placeholder.prev = temp
        self.size-=1

    # #insert(index, val)
    def insert(self, index, val):
        placeholder = self.head.next
        for i in range(index - 1):
            placeholder = placeholder.next
            
        placement =  Node(val, placeholder.next, placeholder)
        
        placeholder.next = placement
        placeholder.next.next.prev = placement
        self.size+=1
```

```python
dll = DoubleLinkedList()
dll.add(5)
dll.add(10)
dll.add(15)
dll.prepend(20)
dll.insert(1,100)
dll.remove(1)
```
